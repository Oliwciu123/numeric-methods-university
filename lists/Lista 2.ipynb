# Zadanie 4

```python
A = np.array([[0, 0, 2, 1, 2],
              [0, 1, 0, 2, -1],
              [1, 2, 0, -2, 0],
              [0, 0, 0, -1, 1],
              [0, 1, -1, 1, -1]])

b = np.array([1, 1, -4, -2, -1])
```

```python
def gauss_elimination_with_partial_pivoting(A, b):
    # Tworzymy kopie macierzy A i wektora b, aby nie zmieniać oryginałów
    A = A.astype(float)
    b = b.astype(float)
    n = len(b)
    
    # Eliminacja w przód z częściowym wyborem elementu głównego
    for i in range(n):
        # Znajdź wiersz z największym elementem w kolumnie i zamień go z bieżącym wierszem
        max_row = np.argmax(abs(A[i:n, i])) + i
        if A[max_row, i] == 0:
            raise ValueError("Nie ma jednoznacznego rowiązania!")  # Jeśli wartość główna jest zerem, macierz nie jest odwracalna
        A[[i, max_row]] = A[[max_row, i]]  # Zamiana wierszy w macierzy A
        b[[i, max_row]] = b[[max_row, i]]  # Zamiana wierszy w wektorze b

        # Eliminacja wszystkich elementów poniżej elementu głównego
        for j in range(i + 1, n):
            factor = A[j, i] / A[i, i]  # Wyliczenie współczynnika do eliminacji
            A[j, i:] -= factor * A[i, i:]  # Odejmowanie wiersza i od wiersza j, aby wyeliminować element A[j, i]
            b[j] -= factor * b[i]  # Aktualizacja wartości wektora b

    # Podstawienie wstecz 
    x = np.zeros(n)  # Inicjalizacja wektora wynikowego x zerami
    for i in range(n - 1, -1, -1):
        x[i] = (b[i] - np.dot(A[i, i+1:], x[i+1:])) / A[i, i]  # Obliczenie wartości x[i]

    return x


x_gauss = gauss_elimination_with_partial_pivoting(A, b)
x_gauss

```

```python
x_scipy = solve(A, b)
x_scipy
```

# Zadanie 8

```python
def gauss_jordan_inverse(A):
    n = len(A)
    A = A.astype(float)  # Przekształcenie macierzy A do formatu zmiennoprzecinkowego
    
    # Tworzenie rozszerzonej macierzy [A | I]
    I = np.eye(n)  # Macierz jednostkowa o wymiarze n x n
    augmented_matrix = np.hstack((A, I))  # Łączenie macierzy A i I w jedną macierz rozszerzoną
    
    # Przeprowadzanie eliminacji Gaussa-Jordana
    for i in range(n):
        # Znalezienie elementu głównego i ewentualna zamiana wierszy
        max_row = np.argmax(abs(augmented_matrix[i:, i])) + i  # Znalezienie wiersza z największym elementem w kolumnie i
        augmented_matrix[[i, max_row]] = augmented_matrix[[max_row, i]]  # Zamiana bieżącego wiersza z wierszem zawierającym największy element

        # Normalizacja wiersza głównego
        pivot = augmented_matrix[i, i]  # Wybór elementu głównego (pivot)
        augmented_matrix[i] = augmented_matrix[i] / pivot  # Podzielenie całego wiersza przez element główny, aby element główny stał się równy 1

        # Eliminacja bieżącej kolumny we wszystkich wierszach poza wierszem głównym
        for j in range(n):
            if j != i:
                factor = augmented_matrix[j, i]  # Współczynnik eliminacji
                augmented_matrix[j] -= factor * augmented_matrix[i]  # Eliminacja elementu w kolumnie i w wierszu j
    
    # Prawa strona rozszerzonej macierzy jest teraz odwrotnością macierzy A
    A_inv = augmented_matrix[:, n:]  # Wyciągnięcie prawej strony rozszerzonej macierzy
    return A_inv
```

```python
A_8 = np.array([[2, -1, 0, 0, 0, 0],
                [-1, 2, -1, 0, 0, 0],
                [0, -1, 2, -1, 0, 0],
                [0, 0, -1, 2, -1, 0],
                [0, 0, 0, -1, 2, -1],
                [0, 0, 0, 0, -1, 5]])


A_inv_8 = gauss_jordan_inverse(A_8)
A_inv_8
```

```python
A_inv_scipy = inv(A_8)
A_inv_scipy
```

Macierz $A^{-1}$ jest symetryczna, lecz nie jest trójdiagonalna. Oznacza to, że macierz odwrotna do macierzy trójdiagonalnej nie musi być macierzą trójdiagonalną.
