```python
import numpy as np
import matplotlib.pyplot as plt
import cProfile
import seaborn
```

# Zadanie 1.


```python

folder_path = 'C:\\Users\\oliwi\\Desktop\\Metody numeryczne\\'


# Definiowanie funkcji Padé
def z1(x):
    return (6 - 2 * x) / (6 + 4 * x + x ** 2)

def z2(x):
    return (6 - 4 * x + x ** 2) / (6 + 2 * x)

# Definiowanie funkcji exp(-x)
def exp_approx(x):
    return np.exp(-x)

# Generowanie wartości x
x = np.linspace(0, 5, 500)

# Obliczanie wartości funkcji z1, z2 oraz exp(-x)
z1_values = z1(x)
z2_values = z2(x)
exp_values = exp_approx(x)

# Wykres 1: Przybliżenia Padégo funkcji exp(-x)
plt.figure(figsize=(10, 6))
plt.plot(x, exp_values, label='exp(-x)', color='black', linewidth=2)
plt.plot(x, z1_values, label='Padé z1', linestyle='--', color='blue', linewidth=2)
plt.plot(x, z2_values, label='Padé z2', linestyle='-.', color='red', linewidth=2)
plt.title('Przybliżenia Padégo funkcji exp(-x)')
plt.xlabel('x')
plt.ylabel('y')
plt.legend()
plt.grid(True)
plt.savefig(folder_path + 'złączone.png')

# Wykres 2: Oddzielne podwykresy z1 i z2
plt.figure(figsize=(10, 6))

plt.subplot(1, 2, 1)
plt.plot(x, exp_values, label='exp(-x)', color='black', linewidth=2)
plt.plot(x, z1_values, label='Padé z1', linestyle='--', color='blue', linewidth=2)
plt.title('Przybliżenie Padé z1')
plt.xlabel('x')
plt.ylabel('y')
plt.legend()
plt.grid(True)

plt.subplot(1, 2, 2)
plt.plot(x, exp_values, label='exp(-x)', color='black', linewidth=2)
plt.plot(x, z2_values, label='Padé z2', linestyle='-.', color='red', linewidth=2)
plt.title('Przybliżenie Padé z2')
plt.xlabel('x')
plt.ylabel('y')
plt.legend()
plt.grid(True)

plt.tight_layout()
plt.savefig(folder_path + 'wielowykresowe.png')

# Wykres 3: Błędy przybliżeń
error_z1 = np.abs(exp_values - z1_values)
error_z2 = np.abs(exp_values - z2_values)

plt.figure(figsize=(10, 6))
plt.plot(x, error_z1, label='Błąd Padé z1', linestyle='--', color='blue', linewidth=2)
plt.plot(x, error_z2, label='Błąd Padé z2', linestyle='-.', color='red', linewidth=2)
plt.title('Błąd przybliżeń Padégo')
plt.xlabel('x')
plt.ylabel('Błąd')
plt.legend()
plt.grid(True)
plt.savefig(folder_path + 'błąd.png')

```

Przybliżenie Padé dla z1 lepiej odwzorowuje funkcję \( \exp(-x) \) na całym przedziale, natomiast z2 ma większy błąd, zwłaszcza dla większych wartości \( x \). Błąd przybliżenia z2 rośnie znacznie szybciej niż z1, co sprawia, że z1 jest bardziej dokładne.

# Zadanie 2.


```python
def machine_epsilon():
    epsilon = 1.0
    while (1.0 + epsilon) != 1.0:
        epsilon /= 2.0
    return epsilon 

# Uruchomienie programu
epsilon = machine_epsilon()
print(f'Dokładność maszynowa na tym komputerze to: {epsilon}')
```

# Zadanie 5.

```python

def polynomial(x):
    return 6 * x**4 + 5 * x**3 - 13 * x**2 + x + 1
def evaluate_polynomial():
    dx = 0.0001
    x_values = np.arange(-10, 10, dx)
    results = [polynomial(x) for x in x_values]
    return results
cProfile.run('evaluate_polynomial()')
```

# Zadanie 6

```python
def horner_polynomial(x):
    return (((6 * x + 5) * x - 13) * x + 1) * x + 1
def evaluate_horner_polynomial():
    dx = 0.0001
    x_values = np.arange(-10, 10, dx)
    results = [horner_polynomial(x) for x in x_values]
    return results
cProfile.run('evaluate_horner_polynomial()')
```

Schemat Hornera zdecydowanie poprawił wydajność programu. Dzięki temu udało się skrócić całkowity czas wykonania o 38%, a czas wykonywania samej funkcji wielomianu o 46%. Jak można zauważyc redukuja liczby operacji mnożeń i dodawań przy dużej liczbie wywołań (jak w tym przykładzie) przekłada się na oszczędność czasu.
