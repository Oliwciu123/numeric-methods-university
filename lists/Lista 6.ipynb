```python

# Dane z zadania
m = 2000  # masa samochodu [kg]
v = np.array([0, 1.0, 1.8, 2.4, 3.5, 4.4, 5.1, 6.0])  # prędkość [m/s]
P_kW = np.array([0, 4.7, 12.2, 19.0, 31.8, 40.1, 43.8, 43.2])  # moc [kW]
P_W = P_kW * 1000  # moc w W

# Obliczanie funkcji podcałkowej f(v) = v / P(v) dla v > 0 (pomijając v = 0)
v_positive = v[1:]  # od v = 1 m/s
P_positive = P_W[1:]  # odpowiadające wartości P
f_v = v_positive / P_positive

# Ustawienia dla metody Simpsona
a = 1  # dolna granica całkowania [m/s]
b = 6  # górna granica całkowania [m/s]
n = len(v_positive) - 1  # liczba przedziałów (n = liczba punktów - 1)
h = (b - a) / n  # szerokość przedziałów

# Zastosowanie wzoru Simpsona
integral = (h / 3) * (f_v[0] + 4 * sum(f_v[1:-1:2]) + 2 * sum(f_v[2:-2:2]) + f_v[-1])

# Obliczenie czasu Δt
delta_t = m * integral

delta_t
```

```python
import math

# Definicja funkcji podcałkowej
def integrand(x):
    return math.cos(2 * math.acos(x))

# Zakres całkowania
a = -1
b = 1

# Funkcja do obliczania całki metodą Simpsona dla n węzłów
def simpsons_rule(n):
    h = (b - a) / (n - 1)  # szerokość przedziału
    x_values = np.linspace(a, b, n)  # węzły
    f_values = np.array([integrand(x) for x in x_values])  # wartości funkcji w węzłach

    # Zastosowanie wzoru Simpsona
    if n == 3:
        integral = (h / 3) * (f_values[0] + 4 * f_values[1] + f_values[2])
    elif n == 5:
        integral = (h / 3) * (
            f_values[0]
            + 4 * (f_values[1] + f_values[3])
            + 2 * f_values[2]
            + f_values[4]
        )
    elif n == 7:
        integral = (h / 3) * (
            f_values[0]
            + 4 * (f_values[1] + f_values[3] + f_values[5])
            + 2 * (f_values[2] + f_values[4])
            + f_values[6]
        )
    else:
        raise ValueError("Wzór Simpsona jest obsługiwany tylko dla n = 3, 5, 7.")
    return integral

# Obliczanie całki dla 3, 5 i 7 węzłów
integral_3_nodes = simpsons_rule(3)
integral_5_nodes = simpsons_rule(5)
integral_7_nodes = simpsons_rule(7)

integral_3_nodes, integral_5_nodes, integral_7_nodes
```

$\cos(2\theta) = 2\cos^2(\theta) - 1.$

## Podstawienie w naszej funkcji
W funkcji $\cos(2 \cos^{-1}(x))$, oznaczmy $\theta = \cos^{-1}(x)$, czyli $\cos(\theta) = x$. Stąd:
$\cos(2 \cos^{-1}(x)) = \cos(2\theta).$

## Zastosowanie tożsamości podwójnego kąta
Z tożsamości $\cos(2\theta) = 2\cos^2(\theta) - 1$ mamy:
$\cos(2 \cos^{-1}(x)) = 2\cos^2(\theta) - 1.$

## Wyrażenie $\cos^2(\theta)$
Ponieważ $\cos(\theta) = x$, to $\cos^2(\theta) = x^2$.

## Ostateczne podstawienie
Podstawiając $\cos^2(\theta) = x^2$ do wzoru, otrzymujemy:
$\cos(2 \cos^{-1}(x)) = 2x^2 - 1.$

# Zadanie 3

Wykonujemy podstawienie $x^3 = \frac{1}{t}$. A zatem:

$t = \frac{1}{x^3}, \quad x = t^{-\frac{1}{3}}, \quad dx = -\frac{t^{-\frac{4}{3}}}{3} \, dt$

W miarę jak $x$ dąży do nieskończoności, $t$ dąży do zera. Natomiast gdy $x$ dąży do 1, to $t$ również. Całkę po podstawieniu możemy zapisać zatem jako:

$\int_{1}^{\infty} \frac{1}{1 + x^4} \, dx = \int_{0}^{1} \frac{\frac{1}{3} t^{-\frac{4}{3}}}{1 + t^{-\frac{4}{3}}} \, dt$


```python

# Funkcja podcałkowa po podstawieniu x^3 = 1/t
def integrand(t):
    return (1 / 3) * (t**(-4/3)) / (1 + t**(-4/3))

# Parametry
n = 6  # liczba węzłów
epsilon = 1e-6  # minimalna wartość t, aby uniknąć nieskończoności
t_values = np.linspace(epsilon, 1, n)  # węzły
h = (t_values[-1] - t_values[0]) / (n - 1)  # szerokość przedziału

# Zastosowanie wzoru trapezów
f_values = np.array([integrand(t) for t in t_values])
integral_trapezoidal = (h / 2) * (f_values[0] + 2 * sum(f_values[1:-1]) + f_values[-1])

print("Wynik metody trapezów:", integral_trapezoidal)
```

# Zadanie 4

```python

# Funkcja podcałkowa h(θ0)
def h_integrand(theta, theta0):
    return 1 / np.sqrt(1 - (np.sin(theta0 / 2) ** 2) * (np.sin(theta) ** 2))

# Metoda Simpsona
def simpsons_rule_h(theta0, n=100):
    # Zakres całkowania od 0 do π/2
    a = 0
    b = np.pi / 2
    h = (b - a) / n

    # Węzły i wartości funkcji w węzłach
    theta_values = np.linspace(a, b, n + 1)
    f_values = np.array([h_integrand(theta, theta0) for theta in theta_values])

    # Zastosowanie wzoru Simpsona
    integral = (h / 3) * (
        f_values[0] + 4 * sum(f_values[1:-1:2]) + 2 * sum(f_values[2:-2:2]) + f_values[-1]
    )
    return integral

# Wartości θ0 w stopniach (zamiana na radiany)
theta0_values_deg = [0,15, 30, 45]
theta0_values_rad = [np.radians(theta) for theta in theta0_values_deg]

# Obliczenie h(θ0) dla każdej wartości
h_values = [simpsons_rule_h(theta0) for theta0 in theta0_values_rad]

# Przybliżenie harmoniczne h(0) = π/2
h_harmonic = np.pi / 2

# Wyświetlenie wyników
print("Wyniki obliczeń dla h(θ₀):")
for theta_deg, h in zip(theta0_values_deg, h_values):
    print(f"θ₀ = {theta_deg}°: h(θ₀) = {h:.6f}, różnica od h(0) = {abs(h - h_harmonic):.6f}")

# Wyświetlenie przybliżenia harmonicznego
print(f"\nPrzybliżenie harmoniczne: h(0) = {h_harmonic:.6f}")
```

```python


# Funkcja podcałkowa
def f(t):
    return np.log(t) / (t**2 - 2 * t + 2)

# Parametry przedziału
a, b = 1, np.pi

# Współczynniki i węzły dla Gaussa-Legendre'a
# 2 węzły
nodes_2 = np.array([-np.sqrt(1/3), np.sqrt(1/3)])
weights_2 = np.array([1, 1])

# 4 węzły
nodes_4 = np.array([-0.861136, -0.339981, 0.339981, 0.861136])
weights_4 = np.array([0.347855, 0.652145, 0.652145, 0.347855])

# Transformacja węzłów
def transform(x, a, b):
    return 0.5 * (b - a) * x + 0.5 * (b + a)

# Kwadratura Gaussa-Legendre'a
def gauss_legendre_integral(f, nodes, weights, a, b):
    transformed_nodes = transform(nodes, a, b)  # Transformacja węzłów
    return 0.5 * (b - a) * np.sum(weights * f(transformed_nodes))

# Obliczenia dla 2 i 4 węzłów
result_2 = gauss_legendre_integral(f, nodes_2, weights_2, a, b)
result_4 = gauss_legendre_integral(f, nodes_4, weights_4, a, b)

# Wyniki
result_2, result_4
```

# Zadanie 6

```python
import numpy as np
import pandas as pd

# Definicje funkcji
def f1(x):
    return x**3 - 2*x

def f2(x):
    return np.sin(x)

def f3(x):
    return np.exp(x)

# Funkcje do różnic dzielonych
def Df1(f, x, h):
    return (f(x + h) - f(x)) / h

def Dc2(f, x, h):
    return (f(x + h) - f(x - h)) / (2 * h)

def Dc4(f, x, h):
    return (2 * Dc2(f, x, h) - Dc2(f, x, 2 * h))

# Parametry
points = [1, np.pi / 3, 0]  # Punkty, w których liczymy pochodne
h_values = [0.1, 0.01, 0.001]  # Kroki różnicowe

# Analityczne pochodne
analytical_derivatives = [
    3 * 1**2 - 2,        # f1'(1) = 1
    np.cos(np.pi / 3),   # f2'(π/3) = 1/2
    np.exp(0)            # f3'(0) = 1
]

# Obliczenia błędów aproksymacji
errors = []

for (func, x_val, analytical, func_name) in zip([f1, f2, f3], points, analytical_derivatives, ["f1", "f2", "f3"]):
    for h in h_values:
        # Obliczenia dla każdego operatora
        df1_result = Df1(func, x_val, h)
        dc2_result = Dc2(func, x_val, h)
        dc4_result = Dc4(func, x_val, h)
        
        errors.append({
            "Funkcja": func_name,
            "Punkt": x_val,
            "h": h,
            "f'(x) - Df1": analytical - df1_result,
            "f'(x) - Dc2": analytical - dc2_result,
            "f'(x) - Dc4": analytical - dc4_result
        })

# Tworzenie tabeli z błędami aproksymacji
errors_df = pd.DataFrame(errors)

# Wyświetlenie tabeli w terminalu
print(errors_df.to_string(index=False))
```

```python

# Dane
x_values = [0.0, 0.1, 0.2, 0.3, 0.4]
f_values = [0.0, 0.078348, 0.138910, 0.192916, 0.244981]

# Symboliczne zmienne
x = symbols('x')

# Funkcja do budowy wielomianu interpolacyjnego Lagrange'a
def lagrange_interpolation(x_values, f_values):
    n = len(x_values)
    polynomial = 0

    for i in range(n):
        # Budowanie bazy Lagrange'a L_i(x)
        term = f_values[i]
        for j in range(n):
            if i != j:
                term *= (x - x_values[j]) / (x_values[i] - x_values[j])
        polynomial += term

    return simplify(polynomial)

# Budowanie wielomianu interpolacyjnego metodą Lagrange'a
lagrange_polynomial = lagrange_interpolation(x_values, f_values)

# Obliczanie pochodnej wielomianu
lagrange_polynomial_derivative = diff(lagrange_polynomial, x)

# Obliczenie wartości pochodnej w punkcie x = 0.2
f_prime_at_02_lagrange = lagrange_polynomial_derivative.subs(x, 0.2)

# Wyniki
print("Wielomian interpolacyjny P(x):")
print(lagrange_polynomial)
print("\nPochodna wielomianu P'(x):")
print(lagrange_polynomial_derivative)
print(f"\nPochodna w punkcie x = 0.2: f'(0.2) ≈ {f_prime_at_02_lagrange}")
```
