# Zadanie 1

```python
# Równanie różniczkowe: y' = x^2 - 4y
f = lambda x, y: x**2 - 4 * y

# Rozwiązanie analityczne
exact_solution = lambda x: (31 / 32) * np.exp(-4 * x) + (1 / 4) * x**2 - (1 / 8) * x + (1 / 32)

# Parametry początkowe
x0 = 0
y0 = 1
x_end = 0.03

# Liczba kroków
steps_list = [1, 2, 4]
```

```python

def euler_method(f, x0, y0, x_end, steps):
    """
    Implementacja metody Eulera.

    :param f: Funkcja f(x, y) = y' w równaniu różniczkowym.
    :param x0: Początkowy punkt x.
    :param y0: Początkowy punkt y.
    :param x_end: Koniec przedziału obliczeń.
    :param steps: Liczba kroków czasowych.
    :return: Przybliżona wartość y w x_end.
    """
    h = (x_end - x0) / steps  # krok czasowy
    x = x0
    y = y0

    for _ in range(steps):
        y += h * f(x, y)
        x += h

    return y


# Obliczenia i porównanie
print("Porównanie metody Eulera z rozwiązaniem analitycznym:\n")
for steps in steps_list:
    y_euler = euler_method(f, x0, y0, x_end, steps)
    y_exact = exact_solution(x_end)
    error = abs(y_euler - y_exact)

    print(f"Liczba kroków: {steps}")
    print(f"y (Euler) = {y_euler:.5f}, y (dokładne) = {y_exact:.5f}, błąd = {error:.5e}\n")
```

```python
def runge_kutta_2(f, x0, y0, x_end, steps):
    """
    Implementacja metody Rungego-Kutty drugiego rzędu.

    :param f: Funkcja f(x, y) = y' w równaniu różniczkowym.
    :param x0: Początkowy punkt x.
    :param y0: Początkowy punkt y.
    :param x_end: Koniec przedziału obliczeń.
    :param steps: Liczba kroków czasowych.
    :return: Przybliżona wartość y w x_end.
    """
    h = (x_end - x0) / steps  # krok czasowy
    x = x0
    y = y0

    for _ in range(steps):
        k1 = f(x, y)
        k2 = f(x + h, y + h * k1)
        y += h * (k1 + k2) / 2
        x += h

    return y
# Obliczenia i porównanie
print("Porównanie metody Rungego-Kutty drugiego rzędu z rozwiązaniem analitycznym:\n")
for steps in steps_list:
    y_rk2 = runge_kutta_2(f, x0, y0, x_end, steps)
    y_exact = exact_solution(x_end)
    error = abs(y_rk2 - y_exact)

    print(f"Liczba kroków: {steps}")
    print(f"y (RK2) = {y_rk2:.5f}, y (dokładne) = {y_exact:.5f}, błąd = {error:.5e}\n")
```

```python
def runge_kutta_4(f, x0, y0, x_end, steps):
    """
    Implementacja metody Rungego-Kutty czwartego rzędu.

    :param f: Funkcja f(x, y) = y' w równaniu różniczkowym.
    :param x0: Początkowy punkt x.
    :param y0: Początkowy punkt y.
    :param x_end: Koniec przedziału obliczeń.
    :param steps: Liczba kroków czasowych.
    :return: Przybliżona wartość y w x_end.
    """
    h = (x_end - x0) / steps  # krok czasowy
    x = x0
    y = y0

    for _ in range(steps):
        k1 = f(x, y)
        k2 = f(x + h / 2, y + h * k1 / 2)
        k3 = f(x + h / 2, y + h * k2 / 2)
        k4 = f(x + h, y + h * k3)
        y += h * (k1 + 2 * k2 + 2 * k3 + k4) / 6
        x += h

    return y
# Obliczenia i porównanie
print("Porównanie metody Rungego-Kutty czwartego rzędu z rozwiązaniem analitycznym:\n")
for steps in steps_list:
    y_rk4 = runge_kutta_4(f, x0, y0, x_end, steps)
    y_exact = exact_solution(x_end)
    error = abs(y_rk4 - y_exact)

    print(f"Liczba kroków: {steps}")
    print(f"y (RK4) = {y_rk4:.5f}, y (dokładne) = {y_exact:.5f}, błąd = {error:.5e}\n")
```

```python
def euler_method(f, x0, y0, x_end, h):
    """
    Implementacja metody Eulera.

    :param f: Funkcja f(x, y) = y' w równaniu różniczkowym.
    :param x0: Początkowy punkt x.
    :param y0: Początkowy punkt y.
    :param x_end: Koniec przedziału obliczeń.
    :param h: Krok czasowy.
    :return: Tablice x i y z wynikami.
    """
    steps = int((x_end - x0) / h)
    x = np.linspace(x0, x_end, steps + 1)
    y = np.zeros(steps + 1)
    y[0] = y0

    for i in range(steps):
        y[i + 1] = y[i] + h * f(x[i], y[i])

    return x, y

def runge_kutta_4(f, x0, y0, x_end, h):
    """
    Implementacja metody Rungego-Kutty czwartego rzędu.

    :param f: Funkcja f(x, y) = y' w równaniu różniczkowym.
    :param x0: Początkowy punkt x.
    :param y0: Początkowy punkt y.
    :param x_end: Koniec przedziału obliczeń.
    :param h: Krok czasowy.
    :return: Tablice x i y z wynikami.
    """
    steps = int((x_end - x0) / h)
    x = np.linspace(x0, x_end, steps + 1)
    y = np.zeros(steps + 1)
    y[0] = y0

    for i in range(steps):
        k1 = f(x[i], y[i])
        k2 = f(x[i] + h / 2, y[i] + h * k1 / 2)
        k3 = f(x[i] + h / 2, y[i] + h * k2 / 2)
        k4 = f(x[i] + h, y[i] + h * k3)
        y[i + 1] = y[i] + h * (k1 + 2 * k2 + 2 * k3 + k4) / 6

    return x, y

# Równanie różniczkowe: y' = sin(y), y(0) = 1
f = lambda x, y: np.sin(y)

# Parametry początkowe
x0 = 0
y0 = 1
x_end = 0.5
h = 0.1

# Rozwiązanie numeryczne
x_euler, y_euler = euler_method(f, x0, y0, x_end, h)
x_rk4, y_rk4 = runge_kutta_4(f, x0, y0, x_end, h)

# Rozwiązanie dokładne za pomocą solve_ivp
solution = solve_ivp(f, [x0, x_end], [y0], t_eval=np.linspace(x0, x_end, int((x_end - x0) / h) + 1))
x_exact = solution.t
y_exact = solution.y[0]

# Wykres porównawczy
plt.figure(figsize=(10, 6))
plt.plot(x_euler, y_euler, label="Metoda Eulera", marker='o')
plt.plot(x_rk4, y_rk4, label="Metoda Rungego-Kutty 4 rzędu", marker='s')
plt.plot(x_exact, y_exact, label="Rozwiązanie dokładne", linestyle='--')
plt.xlabel("x")
plt.ylabel("y")
plt.title("Porównanie metod numerycznych z rozwiązaniem dokładnym dla y' = sin(y), y(0) = 1")
plt.legend()
plt.grid()
plt.show()

# Wyświetlenie wartości w punkcie x = 0.5
y_euler_final = y_euler[-1]
y_rk4_final = y_rk4[-1]
y_exact_final = y_exact[-1]

print(f"Wartości w punkcie x = {x_end}:")
print(f"Metoda Eulera: {y_euler_final:.5f}")
print(f"Metoda Rungego-Kutty 4 rzędu: {y_rk4_final:.5f}")
print(f"Rozwiązanie dokładne: {y_exact_final:.5f}")
```

# Zadanie 3

Wprowadźmy dodatkowe zmienne:
$$
y_1 = \theta, \quad y_2 = \frac{d\theta}{d\tau}.
$$

Czyli mamy równania:
$$
\frac{dy_1}{d\tau} = y_2, \quad \frac{dy_2}{d\tau} = -\sin(y_1) - \frac{y_2}{Q} + \hat{A} \cos(\hat{\omega} \tau).
$$

Można je zapisać jako:
$$
\frac{dy_1}{d\tau} = y_2,
$$
$$
\frac{dy_2}{d\tau} = -\sin(y_1) - \frac{1}{Q} y_2 + \hat{A} \cos(\hat{\omega} \tau).
$$

Czyli:
$$
\frac{dy_1}{d\tau} = f(\tau, y_1, y_2), \quad \frac{dy_2}{d\tau} = g(\tau, y_1, y_2),
$$
gdzie:
$$
f(\tau, y_1, y_2) = y_2, \quad g(\tau, y_1, y_2) = -\sin(y_1) - \frac{y_2}{Q} + \hat{A} \cos(\hat{\omega} \tau).
$$

```python


def runge_kutta_4_system(f, y0, t):
    """
    Rozwiązuje układ równań różniczkowych metodą Rungego-Kutty czwartego rzędu.

    :param f: Funkcja zwracająca pochodne [y', y'' ].
    :param y0: Wartości początkowe [theta, v].
    :param t: Tablica czasu.
    :return: Tablica rozwiązań [theta, v].
    """
    n = len(t)
    y = np.zeros((n, len(y0)))
    y[0] = y0

    for i in range(n - 1):
        h = t[i + 1] - t[i]
        k1 = f(t[i], y[i])
        k2 = f(t[i] + h / 2, y[i] + h * k1 / 2)
        k3 = f(t[i] + h / 2, y[i] + h * k2 / 2)
        k4 = f(t[i] + h, y[i] + h * k3)
        y[i + 1] = y[i] + h * (k1 + 2 * k2 + 2 * k3 + k4) / 6

    return y

def pendulum_dynamics(t, y, Q, omega_hat, A_hat):
    """
    Równanie ruchu wahadła matematycznego z tłumieniem i siłą wymuszającą.

    :param t: Czas.
    :param y: Wektor stanu [theta, v].
    :param Q: Współczynnik tłumienia.
    :param omega_hat: Częstotliwość wymuszania.
    :param A_hat: Amplituda wymuszania.
    :return: Wektor pochodnych [theta', v'].
    """
    theta, v = y
    dtheta_dt = v
    dv_dt = -np.sin(theta) - v / Q + A_hat * np.cos(omega_hat * t)
    return np.array([dtheta_dt, dv_dt])

# Parametry
Q = 2
omega_hat = 2 / 3
scenarios = [
    {"A_hat": 0.5, "theta0": 0.01, "v0": 0},
    {"A_hat": 0.5, "theta0": 0.3, "v0": 0},
    {"A_hat": 1.35, "theta0": 0.3, "v0": 0},
]

# Czas
t = np.linspace(0, 100, 10000)

# Obliczenia i wykresy
for idx, scenario in enumerate(scenarios):
    A_hat = scenario["A_hat"]
    theta0 = scenario["theta0"]
    v0 = scenario["v0"]

    # Rozwiązanie układu równań
    y0 = [theta0, v0]
    results = runge_kutta_4_system(
        lambda t, y: pendulum_dynamics(t, y, Q, omega_hat, A_hat), y0, t
    )
    theta = results[:, 0]
    v = results[:, 1]

    # Wykres theta(t)
    plt.figure(figsize=(8, 5))
    plt.plot(t, theta, label=f"A_hat={A_hat}, theta0={theta0}")
    plt.xlabel("Czas (t)")
    plt.ylabel("Przemieszczenie kątowe (theta)")
    plt.title(f"Przemieszczenie kątowe theta(t), scenariusz {idx + 1}")
    plt.legend()
    plt.grid()
    plt.show()

    # Portret fazowy
    plt.figure(figsize=(8, 5))
    plt.plot(theta, v, label=f"A_hat={A_hat}, theta0={theta0}")
    plt.xlabel("$\\theta(t)$")
    plt.ylabel("$\\frac{d\\theta}{dt}$")
    plt.title(f"Portret fazowy, scenariusz {idx + 1}")
    plt.legend()
    plt.grid()
    plt.show()
```

# Zadanie 4

Rozważamy ruch piłki rzuconej ukośnie, zaniedbując opory powietrza:

- **Rozkład prędkości początkowej na składowe**:  
  Prędkość początkowa $v_0$ jest rozkładana na składową poziomą $v_{0x}$ i pionową $v_{0y}$:
  $v_{0x} = v_0 \cos\theta, \quad v_{0y} = v_0 \sin\theta,$  
  gdzie $\theta$ to kąt wyrzutu piłki względem poziomu.

- **Położenie w kierunku poziomym**:  
  W ruchu poziomym prędkość $v_{0x}$ jest stała, a położenie $x$ w danej chwili $t$ wynosi:
  $x(t) = v_{0x} t = v_0 \cos\theta \cdot t.$

- **Położenie w kierunku pionowym**:  
  W ruchu pionowym położenie $y$ w danej chwili $t$ wynosi:
  $y(t) = v_{0y} t - \frac{1}{2} g t^2 = v_0 \sin\theta \cdot t - \frac{1}{2} g t^2.$

- **Czas całkowity lotu**:  
  Czas, w którym piłka wraca na ziemię ($y = 0$), można znaleźć z równania:
  $t_{\text{total}} = \frac{2 v_0 \sin\theta}{g}.$

- **Zasięg maksymalny**:  
  Maksymalna odległość, jaką osiąga piłka w kierunku poziomym ($x_{\text{max}}$), wynosi:
  $x_{\text{max}} = v_0 \cos\theta \cdot t_{\text{total}} = \frac{v_0^2 \sin(2\theta)}{g}.$

- **Maksymalna wysokość**:  
  Maksymalna wysokość, jaką osiąga piłka ($y_{\text{max}}$), wynosi:
  $y_{\text{max}} = \frac{v_0^2 \sin^2\theta}{2g}.$


```python

# Parametry początkowe
v0 = 20  # prędkość początkowa w m/s
theta_deg = 45  # kąt wyrzutu w stopniach
g = 9.81  # przyspieszenie ziemskie w m/s^2

# Przekształcenie kąta na radiany
theta = np.radians(theta_deg)

# Czas lotu
t_total = 2 * v0 * np.sin(theta) / g

# Wektory czasu
t = np.linspace(0, t_total, 500)

# Równania ruchu
x = v0 * np.cos(theta) * t
y = v0 * np.sin(theta) * t - 0.5 * g * t**2

# Wykres trajektorii
plt.figure(figsize=(10, 6))
plt.plot(x, y, label=f"v0={v0} m/s, θ={theta_deg}°")
plt.axhline(0, color="black", linewidth=0.5, linestyle="--")  # linia pozioma
plt.xlabel("Odległość pozioma (x) [m]")
plt.ylabel("Wysokość (y) [m]")
plt.title("Trajektoria piłki rzuconej ukośnie (bez oporu powietrza)")
plt.legend()
plt.grid()
plt.show()
```

Rozważamy układ równań różniczkowych pierwszego rzędu opisujących dynamikę ruchu piłki:

1. **Położenie**:  
   Równania opisujące położenie w kierunkach \(x\) i \(y\):
   $$
   \frac{dx}{dt} = v_x, \quad \frac{dy}{dt} = v_y.
   $$

2. **Przyspieszenie**:  
   Przyspieszenie w kierunku \(x\):
   $$
   F_x = -\frac{1}{2} C_w \rho A |\vec{V}| v_x,
   $$
   $$
   \frac{dv_x}{dt} = \frac{F_x}{m} = -\frac{1}{2} \frac{C_w \rho A |\vec{V}| v_x}{m},
   $$
   gdzie:
   $$
   |\vec{V}| = \sqrt{v_x^2 + v_y^2}.
   $$

   Przyspieszenie w kierunku \(y\):
   $$
   F_y = -\frac{1}{2} C_w \rho A |\vec{V}| v_y - m \cdot g,
   $$
   $$
   \frac{dv_y}{dt} = \frac{F_y}{m} = -\frac{1}{2} \frac{C_w \rho A |\vec{V}| v_y}{m} - g.
   $$

```python

def runge_kutta_4_system(f, y0, t):
    """
    Rozwiązuje układ równań różniczkowych metodą Rungego-Kutty czwartego rzędu.

    :param f: Funkcja zwracająca pochodne [y', y''].
    :param y0: Wartości początkowe [x, y, vx, vy].
    :param t: Tablica czasu.
    :return: Tablica rozwiązań [x, y, vx, vy].
    """
    n = len(t)
    y = np.zeros((n, len(y0)))
    y[0] = y0

    for i in range(n - 1):
        h = t[i + 1] - t[i]
        k1 = f(t[i], y[i])
        k2 = f(t[i] + h / 2, y[i] + h * k1 / 2)
        k3 = f(t[i] + h / 2, y[i] + h * k2 / 2)
        k4 = f(t[i] + h, y[i] + h * k3)
        y[i + 1] = y[i] + h * (k1 + 2 * k2 + 2 * k3 + k4) / 6

    return y

def projectile_dynamics(t, state, cw, rho, A, m, g):
    """
    Dynamika ruchu piłki z oporem powietrza.

    :param t: Czas.
    :param state: Wektor stanu [x, y, vx, vy].
    :param cw: Współczynnik oporu powietrza.
    :param rho: Gęstość powietrza.
    :param A: Pole przekroju poprzecznego piłki.
    :param m: Masa piłki.
    :param g: Przyspieszenie ziemskie.
    :return: Wektor pochodnych [vx, vy, ax, ay].
    """
    x, y, vx, vy = state
    v = np.sqrt(vx**2 + vy**2)  # prędkość całkowita
    Fx = -0.5 * cw * rho * A * v * vx  # siła oporu w osi x
    Fy = -0.5 * cw * rho * A * v * vy - m * g  # siła oporu w osi y (z grawitacją)
    ax = Fx / m
    ay = Fy / m
    return np.array([vx, vy, ax, ay])

# Parametry początkowe
g = 9.81  # przyspieszenie ziemskie w m/s^2
cw = 0.35  # współczynnik oporu powietrza
rho = 1.2  # gęstość powietrza w kg/m^3

# Scenariusze
sizes = [0.03, 0.045, 0.06]  # różne średnice piłek (pole przekroju poprzecznego)
velocities = [15, 20, 25]  # różne prędkości początkowe
angles = [30, 45, 60]  # różne kąty rzutu

# Osobne wykresy dla każdego rozmiaru piłki
for i, A in enumerate([np.pi * (d/2)**2 for d in sizes]):
    plt.figure(figsize=(10, 6))
    for v0 in velocities:
        for theta_deg in angles:
            theta = np.radians(theta_deg)
            m = 0.145  # masa piłki w kg

            # Warunki początkowe
            x0, y0 = 0, 0
            vx0 = v0 * np.cos(theta)
            vy0 = v0 * np.sin(theta)
            state0 = [x0, y0, vx0, vy0]

            # Czas symulacji
            t_total = 5  # maksymalny czas symulacji w sekundach
            t = np.linspace(0, t_total, 1000)

            # Rozwiązanie układu równań
            results = runge_kutta_4_system(
                lambda t, state: projectile_dynamics(t, state, cw, rho, A, m, g), state0, t
            )

            # Wyodrębnienie wyników
            x = results[:, 0]
            y = results[:, 1]

            # Filtracja punktów poniżej ziemi
            y[y < 0] = np.nan

            # Wykres trajektorii
            plt.plot(x, y, label=f"v0={v0}m/s, θ={theta_deg}°")

    plt.axhline(0, color="black", linewidth=0.5, linestyle="--")  # linia pozioma
    plt.xlabel("Odległość pozioma (x) [m]")
    plt.ylabel("Wysokość (y) [m]")
    plt.title(f"Trajektorie piłki (średnica={sizes[i]} m)")
    plt.legend()
    plt.grid()
    plt.show()
```

# Zadanie 5

Przekształcenie równania drugiego rzędu na układ równań pierwszego rzędu:

- Wprowadź zmienne pomocnicze:
$
u_1 = y, \quad u_2 = \frac{dy}{dx}.
$

- Układ równań przyjmuje postać:
$
u_1' = u_2, \quad u_2' = -\left(1 - 0.2x\right)u_1^2.
$


```python

def runge_kutta_4_step(f, x, y, h):
    """
    Wykonuje jeden krok metody Rungego-Kutty czwartego rzędu.

    :param f: Funkcja definiująca układ równań różniczkowych.
    :param x: Aktualna wartość x.
    :param y: Aktualna wartość wektora y.
    :param h: Krok całkowania.
    :return: Wartość y po kroku h.
    """
    k1 = f(x, y)
    k2 = f(x + h / 2, y + h * k1 / 2)
    k3 = f(x + h / 2, y + h * k2 / 2)
    k4 = f(x + h, y + h * k3)
    return y + h * (k1 + 2 * k2 + 2 * k3 + k4) / 6

def solve_runge_kutta_4(f, a, b, y0, n):
    """
    Rozwiązuje układ równań różniczkowych metodą Rungego-Kutty czwartego rzędu.

    :param f: Funkcja definiująca układ równań różniczkowych.
    :param a: Początek przedziału.
    :param b: Koniec przedziału.
    :param y0: Warunki początkowe.
    :param n: Liczba kroków całkowania.
    :return: Tablice x i y z wynikami.
    """
    x = np.linspace(a, b, n + 1)
    y = np.zeros((n + 1, len(y0)))
    y[0] = y0
    h = (b - a) / n

    for i in range(n):
        y[i + 1] = runge_kutta_4_step(f, x[i], y[i], h)

    return x, y

def shooting_method_rk4(f, a, b, ya, yb, s_guess, tol=1e-6, max_iter=100, n=100):
    """
    Rozwiązuje zagadnienie brzegowe metodą strzałów z użyciem metody Rungego-Kutty 4 rzędu.

    :param f: Funkcja definiująca układ równań różniczkowych.
    :param a: Początek przedziału.
    :param b: Koniec przedziału.
    :param ya: Wartość y(a).
    :param yb: Wartość y(b).
    :param s_guess: Początkowe przybliżenie dla pochodnej początkowej (s).
    :param tol: Tolerancja błędu.
    :param max_iter: Maksymalna liczba iteracji.
    :param n: Liczba kroków całkowania.
    :return: Tablica rozwiązania (x, y).
    """
    def solve_ivp_with_s(s):
        """Rozwiązuje układ równań dla danego s."""
        def ode_system(x, y):
            y1, y2 = y
            dy1_dx = y2
            dy2_dx = -(1 - 0.2 * x) * y1**2
            return np.array([dy1_dx, dy2_dx])

        return solve_runge_kutta_4(ode_system, a, b, [ya, s], n)

    s_low, s_high = s_guess - 1, s_guess + 1
    for _ in range(max_iter):
        # Rozwiązanie dla dolnego i górnego strzału
        x_low, y_low = solve_ivp_with_s(s_low)
        x_high, y_high = solve_ivp_with_s(s_high)

        # Sprawdzenie warunku brzegowego
        yb_low = y_low[-1, 0]  # y(b) dla s_low
        yb_high = y_high[-1, 0]  # y(b) dla s_high

        if abs(yb_low - yb) < tol:
            return x_low, y_low
        if abs(yb_high - yb) < tol:
            return x_high, y_high

        # Aktualizacja s przez interpolację liniową (metoda strzałów z bisekcją)
        s_new = s_low + (s_high - s_low) * (yb - yb_low) / (yb_high - yb_low)
        x_new, y_new = solve_ivp_with_s(s_new)
        yb_new = y_new[-1, 0]

        if abs(yb_new - yb) < tol:
            return x_new, y_new

        # Aktualizacja przedziału strzałów
        if yb_new < yb:
            s_low, s_high = s_new, s_high
        else:
            s_low, s_high = s_low, s_new

    raise RuntimeError("Nie osiągnięto zbieżności w metodzie strzałów.")

# Parametry problemu
a = 0
b = np.pi / 2
ya = 0
yb = 1
s_guess = 2  # Początkowe przybliżenie dla s
n = 100  # Liczba kroków całkowania

# Rozwiązanie zagadnienia brzegowego
x, y = shooting_method_rk4(
    f=lambda x, y: [y[1], -(1 - 0.2 * x) * y[0]**2],
    a=a,
    b=b,
    ya=ya,
    yb=yb,
    s_guess=s_guess,
    n=n
)

# Wyświetlenie wyników
y1 = y[:, 0]
plt.figure(figsize=(10, 6))
plt.plot(x, y1, label="Rozwiązanie numeryczne (RK4)")
plt.xlabel("x")
plt.ylabel("y(x)")
plt.title("Rozwiązanie zagadnienia brzegowego metodą strzałów (RK4)")
plt.legend()
plt.grid()
plt.show()
```
